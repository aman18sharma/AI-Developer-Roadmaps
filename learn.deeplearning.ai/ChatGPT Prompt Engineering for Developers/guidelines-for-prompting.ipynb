````python
import openai
from openai import OpenAI

import os

from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv())


API_KEY = os.getenv("OPENAI_API_KEY")
````

````python
from openai import OpenAI, APIError, APIConnectionError, RateLimitError, BadRequestError

def get_completion(prompt):
    if not prompt or not prompt.strip():
        raise ValueError("Prompt must be a non-empty string.")

    client = OpenAI(
        api_key=os.getenv("XKIRO_OPENAI_API_KEY"),
        base_url=os.getenv("XKIRO_BASE_URL")
    )

    messages = [{"role": "user", "content": prompt}]  # fixed: "content" not "context"

    try:
        response = client.chat.completions.create(
            model=os.getenv("XKIRO_MODEL_NAME"),
            messages=messages
        )
        return response.choices[0].message.content  # fixed: was `choices[0]`

    except BadRequestError as e:
        print(f"Bad request: {e}")
        raise
    except RateLimitError as e:
        print(f"Rate limit exceeded: {e}")
        raise
    except APIConnectionError as e:
        print(f"Connection error: {e}")
        raise
    except APIError as e:
        print(f"API error: {e}")
        raise
    except Exception as e:
        print(f"Unexpected error: {e}")
        raise
````

````python
text = f"""
You should express what you want a model to do by providing instructions that are as clear and specific as you can possibly make them. \
This will guide the model towards the desired output, and reduce the chances of receiving irrelevant or incorrect responses. \
Don't confuse writing a clear prompt with writing a short prompt. In many cases, longer prompts provide more clarity and context for \
the model, which can lead to more detailed and relevant outputs.
"""

prompt = f"""
Summarize the text delimited by triple backticks into a single sentence.
```{text}```
"""

try:
    response = get_completion(prompt)
    print(response)
except Exception as e:
    print(f"Failed to get completion: {e}")
````

````python
prompt = f"""Generate a list of three made-up book titles along with their authors and genres. \
Provide then in JSON format with the following keys: book_id, title, author, genre.
"""

response = get_completion(prompt)
print(response)
````
